```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Matriz de caracteristicas (4 casas, 3 rasgos reales)
# Agregamos una columna redundante (m2 * 10) para simular ruido
X_raw = np.array([
    [80, 3, 10],
    [120, 3, 5],
    [100, 2, 20],
    [150, 4, 2]
])

# Precios reales
y = np.array([310, 390, 325, 530])

# 2. Preparacion: Añadir Bias (Columna de unos para Theta 0)
X = np.c_[np.ones((X_raw.shape[0], 1)), X_raw]
print(f"Dataset X (con Bias) preparado. shape {X.shape}")
```

```python
# MAtriz de Gram (A)
A = X.T @ X

# Diagnostico
rango = np.linalg.matrix_rank(A)
det = np.linalg.det(A)

print(f"Rango de la matriz: {rango}")
print(f"Determinante: {det:.2f}")

if rango < X.shape[1] or np.isclose(det, 0):
    print("Alerta: Matriz singular o redundante detectada")
```

```python
# Calculamos la Pseudoinversa de A
A_pinv = np.linalg.pinv(A)

# Resolevmos para theta (pasos del modelo)
theta = A_pinv @ (X.T @ y)

nombres_weights = ["Intercepto", "m2", "Habitaciones", "Antigüedad"]
for nombre, valor in zip(nombres_weights, theta):
    print(f"{nombre}: {valor:.2f}")
```

```python
# Nueva casa: [Bias=1, m2=110, Hab=3, Edad=8]
nueva_casa = np.array([1, 110, 3, 8])

# Predicción (Producto Punto)
precio_estimado = nueva_casa @ theta

print(f"--- Resultado de Consultoría ---")
print(f"Precio estimado para la propiedad: ${precio_estimado:.2f}k")
```

```python
# Calculamos predicciones para el dataset original
y_pred = X @ theta

# Vector de error (Residuos)
error_vec = y - y_pred

# Magnitud del error total (Norma L2)
error_total = np.linalg.norm(error_vec)

print(f"Error residual total (Norma L2): {error_total:.2f}")
```
